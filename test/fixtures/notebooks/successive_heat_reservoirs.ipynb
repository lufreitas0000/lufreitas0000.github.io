### Precomputed Notebook Fixture: Successive Heat Reservoirs ($N$-Step Entropy Production)
Reference: Lucas R. D. Freitas and Luiz Felipe C. Pereira, *Revista Brasileira de Ensino de Física* **41**(4), e20180331 (2019), [DOI: 10.1590/1806-9126-RBEF-2018-0331](https://doi.org/10.1590/1806-9126-rbef-2018-0331).

For an object of constant heat capacity $C$ brought from $T_0$ to $T_N$ via $N$ equal-step intermediate reservoirs $T_k = T_0 + k\,\Delta T$ with $\Delta T = (T_N - T_0)/N$, the dimensionless total entropy variation is:
$$\frac{\Delta S_{\text{tot}}(N)}{C} = \sum_{k=1}^{N} \left( \frac{T_{k-1}}{T_k} - 1 + \ln\frac{T_k}{T_{k-1}} \right).$$

```python
import math

def delta_s_over_c(t_initial: float, t_final: float, n_reservoirs: int) -> float:
    dt = (t_final - t_initial) / n_reservoirs
    total = 0.0
    for k in range(1, n_reservoirs + 1):
        t_prev = t_initial + (k - 1) * dt
        t_curr = t_initial + k * dt
        total += (t_prev / t_curr) - 1.0 + math.log(t_curr / t_prev)
    return total

for n in (1, 2, 4, 10, 100, 1000):
    print(f"N = {n:4d} -> Delta S_tot / C = {delta_s_over_c(300.0, 600.0, n):.6f}")
```
